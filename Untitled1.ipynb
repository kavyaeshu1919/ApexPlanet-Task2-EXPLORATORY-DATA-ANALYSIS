```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel("/content/Cleaned_Dataset.xlsx")
df.head()
```

```python
plt.figure(figsize=(6,4))
plt.hist(df["Age"], bins=10)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
df["Product"].value_counts().plot(kind="bar", figsize=(8,4))
plt.title("Products Sold")
plt.show()
```

```python
df.groupby("Category")["Total_Sales"].sum().plot(kind="bar", figsize=(7,4))
plt.title("Total Sales by Category")
plt.ylabel("Sales")
plt.show()
```

```python
df.groupby("City")["Total_Sales"].sum().sort_values(ascending=False).head(10).plot(kind="bar")
plt.title("Top Cities by Sales")
plt.show()
```
